```python
import os
from dotenv import load_dotenv
load_dotenv()

openai_api_key = os.getenv("OPENAI_API_KEY")
```

```python
from langchain_community.document_loaders import WebBaseLoader
loader = WebBaseLoader("https://www.educosys.com/course/genai")

docs = loader.load()
print(docs)
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
splits = text_splitter.split_documents(docs)
```

```python
print(splits[0])
```

```python
print(len(splits))
```

```python
# 2️⃣ Generate embeddings for each chunk
from sentence_transformers import SentenceTransformer

# Load model
model = SentenceTransformer('all-MiniLM-L6-v2')

# Extract the text from LangChain Document objects
texts = [doc.page_content for doc in splits]

# Generate embeddings
embeddings = model.encode(texts)
print(f"Embeddings shape: {len(embeddings)} x {len(embeddings[0])}")
```

```python
# 3️⃣ Insert into ChromaDB
import chromadb

client = chromadb.Client()
collection = client.create_collection(name="genai_course")  # Create a collection

# Create unique IDs for each chunk
ids = [f"chunk_{i}" for i in range(len(splits))]

# Add to Chroma
collection.add(
    documents=texts,
    embeddings=embeddings.tolist(),  # make sure it's a list
    ids=ids
)

print("Inserted chunks into ChromaDB!")
```

```python
# 4️⃣ Optional: Query
query = "What is generative AI?"
query_embedding = model.encode([query])

results = collection.query(
    query_embeddings=query_embedding.tolist(),
    n_results=3
)

print(results)
```
